```python
from ngsolve import *
from ngsolve.webgui import Draw

from netgen.occ import unit_square
```

```python
mesh = Mesh(unit_square.GenerateMesh(maxh=0.02))
Draw (mesh)
```

```python
fes = H1(mesh, order=3, dirichlet="bottom|left|right")
u = fes.TrialFunction()
v = fes.TestFunction()

a = BilinearForm(grad(u)*grad(v)*dx)
f = LinearForm (fes)

a.Assemble()
f.Assemble();
```

```python
gfu = GridFunction(fes)

uD = sin(2*pi*x)
gfu.Set (uD, definedon=mesh.Boundaries("bottom"))
```

```python
ea = { "euler_angles" : [-60,0,105] }
Draw(gfu, **ea, deformation=True, scale=0.3);
```

```python
r = f.vec - a.mat * gfu.vec
gfu.vec.data += a.mat.Inverse(fes.FreeDofs()) * r
```

```python
Draw(gfu, **ea, deformation=True, scale=0.3);
```
